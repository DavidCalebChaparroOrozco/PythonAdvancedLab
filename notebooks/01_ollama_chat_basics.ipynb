```python
import ollama

answer = ollama.chat(
    model = "qwen2.5vl:32b",
    messages = [{"role": "user",
                "content": "Greetings, say hello in one word."}]
)
print(type(answer).__name__)
print(answer.message.content)
```

```python
a1 = ollama.chat(
    model = "qwen2.5vl:32b",
    messages = [{"role": "user",
                "content": "My name is Caleb."}])

a2 = ollama.chat(
    model = "qwen2.5vl:32b",
    messages = [{"role": "user",
                "content": "What is my name?"}])

print(a1.message.content)
print(a2.message.content)
```
